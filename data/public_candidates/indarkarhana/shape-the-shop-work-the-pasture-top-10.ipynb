## Method

This submission is a layered visible-state rule network over a complete farm
programme. It guards worker alignment, purchases, placement, same-turn shed
projection, sale capacity, contested-value ordering, and sequential funding.
It then makes two connected production decisions from current observable
state:

1. Conserved cow/sheep pasture bundles are relabelled toward visible dairy or
   wool demand without changing route geometry, animal count, or sale slots.
2. At a single engine-verified checkpoint, one additional demand-aligned
   animal and one hand are purchased. The hand follows a guarded
   pickup/NORTH/PLACE sequence into an already serviced empty pasture. Every
   geometry, inventory, capacity, and cash mismatch fails closed to the parent.

Farm occupancy is part of the deterministic game trajectory, so using the
latent pasture changes later shop/economic states through ordinary engine
mechanics. The policy reacts only to present observation. It never reads team
identity, rank, seed, replay ID, terminal outcome, future shops/actions,
opponent-private inventory, network data, or accelerator inference.

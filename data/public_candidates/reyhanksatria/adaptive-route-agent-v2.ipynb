```python
from pathlib import Path
import base64
import hashlib
import zlib

# Define expected hash and embedded payload
EXPECTED_ARCHIVE_SHA = "b650a31d091323f2510aede0265937d3193a82a99109eadc8ab39c6e85db278d"
PAYLOAD = ""

# Decode and verify embedded payload
archive = zlib.decompress(base64.b64decode(PAYLOAD))
assert hashlib.sha256(archive).hexdigest() == EXPECTED_ARCHIVE_SHA

# Restore original submission archive
Path("submission.tar.gz").write_bytes(archive)

# Verify final archive
archive_sha = hashlib.sha256(Path("submission.tar.gz").read_bytes()).hexdigest()
assert archive_sha == EXPECTED_ARCHIVE_SHA
print({"agent": "ARA-V2", "archive_sha256": archive_sha})
```
